# Partie 1 — Linopy

Linopy permet de formuler un problème d’optimisation à l’aide de variables, de contraintes et d’une fonction objectif. On commence par un problème jouet, puis on ajoute des dimensions et on termine par un marché électrique constitué de deux zones reliées par une ligne.

<img src="https://github.com/PyPSA/linopy/blob/master/doc/logo.png?raw=true" width="300px" />

[Linopy](https://linopy.readthedocs.io/en/latest/index.html) est une librairie pour formuler, résoudre et analyser des problèmes d'optimisation avec Python.

Avec Linopy, on crée en Python des modèles d'optimisation composés de variables de décision, de contraintes et d'objectifs d'optimisation. On les résout ensuite avec des solveurs.

[Linopy](https://linopy.readthedocs.io/en/latest/index.html) prend en charge de nombreux types de problèmes :

- **Programmation linéaire**
- Programmation en nombres entiers
- Programmation mixte (variables continues et entières)
- Programmation quadratique

## Résoudre un modèle de base

Dans cet exemple, on présente les fonctions de base de la classe `Model` de linopy. On commence par un problème d'optimisation linéaire très simple :

Minimiser :
    $$x + 2y$$
sous les contraintes :
    $$ x \ge 0 $$
    $$y \ge 0 $$
    $$3x + 7y \ge 10 $$
    $$5x + 2y \ge 3 $$

### Initialiser un `Model`

La classe `Model` est au cœur de Linopy. C'est un conteneur qui regroupe toutes les données d'un problème d'optimisation linéaire : variables, contraintes et fonction objectif.

````python
# On importe linopy puis on crée un modèle vide : c'est le conteneur du problème
import linopy

m = linopy.Model()
````

Cette instruction crée un nouvel objet `Model`, que l'on utilise ensuite pour définir le problème d'optimisation.

### Ajouter des variables de décision

Les **variables** sont les inconnues d'un problème d'optimisation : leurs valeurs sont fixées par la résolution. Une variable peut toujours être munie d'une borne inférieure et d'une borne supérieure. Ici, `x` et `y` ont une borne inférieure de zéro (par défaut, une variable n'est pas bornée). En linopy, on ajoute des variables à un `Model` avec la méthode `add_variables()` :

````python
# Deux variables continues x et y, bornées par le bas (x >= 0, y >= 0) ; name sert à les retrouver ensuite
x = m.add_variables(lower=0, name="x")
y = m.add_variables(lower=0, name="y");
````

`x` et `y` sont des variables linopy de la classe `linopy.Variable`. Chacune contient toute l'information qui la définit. Le paramètre `name` est facultatif mais utile pour retrouver les variables plus tard.

````python
# Afficher une variable : on voit son nom, ses bornes et sa dimension
x
````

````python
# Le modèle conserve la liste de toutes ses variables
m.variables
````

````python
# On retrouve une variable par son nom
m.variables["x"]
````

### Ajouter des contraintes

Les **contraintes** sont des égalités ou inégalités qui définissent l'espace *faisable* des variables de décision. Elles comportent un membre de gauche (fonction des variables de décisions) et un membre de droite (constant). La première contrainte à écrire est $3x + 7y \ge 10$, que l'on écrit exactement comme en mathématiques :

````python
# Écrire une contrainte avec les opérateurs >=, <=, == fabrique un objet « contrainte »
# (elle n'est pas encore dans le modèle)
3 * x + 7 * y >= 10
````

On peut aussi mélanger la constante et l'expression des variables, comme ceci :

````python
# Même contrainte, avec la constante à gauche : linopy la déplace automatiquement à droite
3 * x + 7 * y - 10 >= 0
````

... et linopy se charge de séparer l'expression des variables (à gauche) des valeurs constantes (à droite).

La contrainte n'est pas encore rattachée au modèle. On l'ajoute en appelant la fonction `add_constraints()` :

````python
# add_constraints ajoute les contraintes au modèle ; elles reçoivent un nom automatique (con0, con1)
m.add_constraints(3 * x + 7 * y >= 10)
m.add_constraints(5 * x + 2 * y >= 3);
````

````python
# La liste des contraintes du modèle
m.constraints
````

````python
# Une contrainte précise, retrouvée par son nom
m.constraints["con0"]
````

### Ajouter l'objectif

La fonction objectif définit ce que l'on veut optimiser : c'est une fonction des variables que le solveur cherche à minimiser ou maximiser. On la définit pour un `linopy.Model` avec la méthode `add_objective()`. Pour notre exemple :

````python
# Fonction objectif : minimiser x + 2y
m.add_objective(x + 2 * y, sense="min")
````

````python
# Affichage de l'objectif
m.objective
````

On peut minimiser ou maximiser avec linopy. Par défaut, linopy utilise `sense='min'` : il n'est donc pas nécessaire de préciser le sens de l'optimisation. En résumé :

````python
# Résumé du modèle : variables, contraintes, objectif
m
````

### Résoudre le modèle

Une fois le `linopy.Model` défini avec ses variables, ses contraintes et sa fonction objectif, on le résout avec la méthode `solve` :

````python
# Résolution (par défaut, linopy choisit un solveur installé ; ici HiGHS)
m.solve(api='direct')
````

Les solveurs calculent les solutions des modèles d'optimisation. Il en existe un grand nombre, souvent spécialisés dans certains types de problèmes ou d'algorithmes (par exemple linéaire ou non linéaire).

- **exemples libres** : [HiGHS](https://highs.dev), [SCIP](https://www.scipopt.org/), [CBC](https://www.coin-or.org/Cbc/), [GLPK](https://www.gnu.org/software/glpk/), [Ipopt](https://coin-or.github.io/Ipopt/)
- **exemples commerciaux** : [Gurobi](https://www.gurobi.com/), [COPT](https://www.shanshu.ai/copt), [CPLEX](https://www.ibm.com/de-de/analytics/cplex-optimizer), [FICO Xpress](https://www.fico.com/en/products/fico-xpress-optimization)

Les solveurs libres suffisent pour des modèles linopy de plusieurs centaines à plusieurs milliers de variables et de contraintes. Pour des applications plus grandes ou plus complexes, on peut avoir besoin d'un solveur commercial (qui propose souvent des licences académiques gratuites).

Dans ce cours, on utilise HiGHS, déjà présent dans l'environnement du cours `esm-ss-26`.

### Récupérer les résultats de l'optimisation

La solution du problème linéaire est affectée aux variables, dans l'attribut `solution`, sous forme de `xarray.Dataset`.

````python
# Valeur optimale de x
x.solution
````

````python
# Valeur optimale de y
y.solution
````

On peut aussi lire la valeur de l'objectif :

````python
# Valeur optimale de l'objectif
m.objective.value
````

Et les valeurs duales (ou prix implicites) des contraintes du modèle :

````python
# Valeur duale (prix implicite) de la contrainte con0 :
# de combien l'objectif varie si le second membre de con0 augmente d'une unité
m.dual["con0"]
````

Bravo ! Vous avez résolu votre premier modèle linopy.

## Utiliser des coordonnées

C'est ici que la puissance du paquet apparaît.

Linopy est construit autour de l'idée que les variables, et donc les expressions et les contraintes, possèdent des coordonnées. Un objet `Variable` contient en réalité de nombreuses variables réparties sur des dimensions, comme un tableau `numpy` ou un `pandas.DataFrame`.

Supposons que les deux variables `x` et `y` dépendent maintenant du temps `t`, et que l'on modifie le problème ainsi :

Minimiser :
$$\sum_t x_t + 2 y_t$$

sous les contraintes :

$$x_t \ge 0 \qquad \forall t $$
$$y_t \ge 0 \qquad \forall t $$
$$3x_t + 7y_t \ge 10 t \qquad \forall t$$
$$5x_t + 2y_t \ge 3 t \qquad \forall t$$

où `t` parcourt les valeurs de 0 à 9.

Pour formuler ce nouveau problème avec linopy, on commence à nouveau par initialiser un modèle.

````python
# Nouveau modèle vide (on repart de zéro pour la version avec coordonnées)
m = linopy.Model()
````

On définit `x` et `y` avec `add_variables()`, mais en ajoutant cette fois l'argument `coords`. Linopy crée alors automatiquement des variables d'optimisation pour toutes les coordonnées, ici les pas de temps `t`.

````python
# On définit la dimension « time » comme un index pandas nommé (10 pas de temps : 0 à 9)
import pandas as pd

time = pd.Index(range(10), name="time")

# coords=[time] crée UNE variable par pas de temps, pour x comme pour y
x = m.add_variables(
    lower=0,
    coords=[time],
    name="x",
)
y = m.add_variables(lower=0, coords=[time], name="y")
````

````python
# x est maintenant un tableau de 10 variables
x
````

On écrit les contraintes avec la même syntaxe qu'avant, en multipliant le membre de droite par `t`. Attention : les coordonnées du membre de gauche et du membre de droite doivent coïncider.

````python
# Le facteur t sous forme de Series indexée par time (mêmes coordonnées que x et y)
factor = pd.Series(time, index=time)

# Une seule écriture produit 10 contraintes (une par pas de temps)
3 * x + 7 * y >= 10 * factor
````

Il est toujours utile d'écrire les contraintes avant de les ajouter au modèle. Comme elles sont correctes, on peut les ajouter.

````python
# On ajoute les deux familles de contraintes en leur donnant un nom explicite
con1 = m.add_constraints(3 * x + 7 * y >= 10 * factor, name="con1")
con2 = m.add_constraints(5 * x + 2 * y >= 3 * factor, name="con2")
m
````

Pour l'objectif, on utilise la fonction `sum` des `linopy.LinearExpression`. Elle additionne tous les termes de la dimension `time` dans une seule grande expression.

````python
# .sum() additionne sur la dimension time : on obtient une seule expression scalaire
obj = (x + 2 * y).sum()
````

````python
obj
````

````python
# overwrite=True remplace l'objectif précédent
m.add_objective(obj, overwrite=True)
````

On peut alors résoudre :

````python
m.solve()
````

Pour examiner la solution, on peut passer par les variables (par exemple `y.solution`) ou par l'agrégateur `solution` du modèle, qui rassemble la solution de toutes les variables.

````python
# Solution de toutes les variables, sous forme de tableau pandas
m.solution.to_dataframe()
````

Il est parfois utile de tracer la solution :

````python
# Tracé de la solution (une courbe par variable)
m.solution.to_dataframe().plot(grid=True, ylabel="Valeur optimale");
````

Parfait ! Vous savez maintenant créer des variables et des expressions linopy avec des coordonnées. Pour des opérations `linopy` plus avancées, consultez le [guide utilisateur](https://linopy.readthedocs.io/en/latest/user-guide.html).

## Partie A — Une zone de marché, une période

On veut minimiser le coût d'exploitation d'un système électrique qui représente l'Afrique du Sud, sous les limites de production des centrales et la satisfaction de la demande :

\begin{equation}
    \min_{g_s} \sum_s o_s g_s
\end{equation}
sous les contraintes :
\begin{align}
    g_s &\leq G_s \\
    g_s &\geq 0 \\
    \sum_s g_s &= d
\end{align}

où $g_s$ est la production de la technologie $s$ (variable de décision), $o_s$ son coût marginal, $G_s$ sa capacité et $d$ la demande (paramètres).

### A.1 Données fournies

Coûts marginaux (en EUR/MWh), capacités des centrales (en MW) et demande inélastique (en MW) :

````python
marginal_costs = pd.Series([0, 30, 60, 80], index=["Wind", "Coal", "Gas", "Oil"])
marginal_costs
````

````python
capacities = pd.Series([3000, 35000, 8000, 2000], index=["Wind", "Coal", "Gas", "Oil"])
capacities
````

````python
load = 42000
````

### A.2 Créer le modèle

On commence par créer un modèle vide, qui contiendra variables, contraintes et objectif.

````python
# TODO : créer un modèle linopy vide nommé m.
### DEBUT SOLUTION
m = linopy.Model()
### FIN SOLUTION
````

### A.3 Ajouter les variables de dispatch

Les variables de dispatch `g` sont bornées :
$$g_s \leq G_s, \qquad g_s \geq 0$$
On crée une variable par technologie : il suffit de passer les technologies comme coordonnée (`coords`).

````python
# TODO : ajouter les variables de dispatch g au modèle m.
### DEBUT SOLUTION
#  - appeler m.add_variables(...) avec lower=0 (borne inférieure) et upper=capacities (borne supérieure par technologie) ;
#  - coords=[capacities.index] crée une variable par technologie ; name="g" ;
#  - afficher g pour vérifier sa dimension.
g = m.add_variables(lower=0, upper=capacities, coords=[capacities.index], name="g")
g
### FIN SOLUTION
````

### A.4 Définir l'objectif

On ajoute l'objectif : minimiser le coût d'exploitation total
$$\min_{g_s} \sum_s o_s g_s$$

````python
# TODO : ajouter l'objectif au modèle (minimisation du coût d'exploitation).
### DEBUT SOLUTION
#  - multiplier les coûts marginaux (marginal_costs.values) par la variable g ;
#  - m.add_objective(expression, sense="min") puis afficher m.objective.
m.add_objective(marginal_costs.values * g, sense="min")
m.objective
### FIN SOLUTION
````

### A.5 Ajouter la contrainte d'équilibre

La production totale doit égaler la demande :
$$\sum_s g_s = d$$
On nomme la contrainte : ce nom servira à récupérer sa valeur duale.

````python
# TODO : ajouter la contrainte d'équilibre offre-demande.
### DEBUT SOLUTION
#  - g.sum() somme les productions de toutes les technologies ;
#  - m.add_constraints(g.sum() == load, name="energy_balance").
m.add_constraints(g.sum() == load, name="energy_balance")
### FIN SOLUTION
````

### A.6 Résoudre le modèle

````python
# TODO : résoudre le modèle avec m.solve().
### DEBUT SOLUTION
m.solve()
### FIN SOLUTION
````

### A.7 Lire les résultats

Le dispatch optimal (en MW) se lit dans la solution du modèle. Le **prix de marché** se lit dans la valeur duale de la contrainte d'équilibre : c'est le coût supplémentaire d'une unité de demande en plus (cf chapitre 4 du cours).

````python
# TODO : afficher le dispatch optimal : m.solution.to_dataframe().
### DEBUT SOLUTION
m.solution.to_dataframe()
### FIN SOLUTION
````

````python
# TODO : afficher le prix de marché : la valeur duale de la contrainte "energy_balance" (m.dual[...]).
### DEBUT SOLUTION
m.dual["energy_balance"]
### FIN SOLUTION
````

## Partie B — Deux zones de marché reliées par une ligne

On ajoute une dimension spatiale : le problème devient

\begin{equation}
  \min_{g_{i,s}, f_\ell} \sum_{i,s} o_{i,s} g_{i,s}
\end{equation}
sous les contraintes :
\begin{align}
  g_{i,s} &\leq G_{i,s} \\
  g_{i,s} &\geq 0 \\
  \sum_s g_{i,s} - \sum_\ell K_{i\ell} f_\ell &= d_i & \text{KCL} \\
  |f_\ell| &\leq F_\ell & \text{limites de ligne}  \\
  \sum_\ell C_{\ell c} x_\ell f_\ell &= 0 & \text{KVL}
\end{align}

Ici, on relie le système sud-africain précédent à une centrale hydraulique du Mozambique par une seule ligne de transport. Comme une seule ligne ne forme aucun cycle, on peut négliger la KVL.

### B.1 Données fournies (toutes en MW, sauf les coûts en EUR/MWh)

````python
generators = ["Coal", "Wind", "Gas", "Oil", "Hydro"]
countries = ["South_Africa", "Mozambique"]
````

````python
capacities = pd.DataFrame(
    {
        "Coal": [35000, 0],
        "Wind": [3000, 0],
        "Gas": [8000, 0],
        "Oil": [2000, 0],
        "Hydro": [0, 1200],
    },
    index=countries,
)
capacities.index.name = "countries"
capacities.columns.name = "generators"

capacities
````

````python
# variable costs in EUR/MWh
marginal_costs = pd.Series([30, 0, 60, 80, 0], index=generators)
marginal_costs.index.name = "generators"
marginal_costs
````

````python
load = pd.Series([42000, 650], index=countries)
load.index.name = "countries"
load
````

````python
transmission = 500
````

### B.2 Créer un nouveau modèle

On repart d'un modèle vide.

````python
# TODO : créer un nouveau modèle linopy vide nommé m (linopy.Model()).
### DEBUT SOLUTION
m = linopy.Model()
### FIN SOLUTION
````

### B.3 Variables de dispatch pour chaque pays et chaque technologie

Cette fois, `capacities` est un `DataFrame` (pays x technologies) : en le passant comme borne `upper`, linopy crée automatiquement une variable par couple (pays, technologie), avec les mêmes dimensions. Regardez d'abord les capacités, puis créez les variables.

````python
# TODO : afficher le tableau des capacités (DataFrame pays x technologies).
### DEBUT SOLUTION
capacities
### FIN SOLUTION
````

````python
# TODO : créer les variables de dispatch g avec lower=0 et upper=capacities (le DataFrame).
#  - donner name="g" ; afficher g : vérifier qu'il a deux dimensions (countries, generators).
### DEBUT SOLUTION
g = m.add_variables(lower=0, upper=capacities, name="g")
g
### FIN SOLUTION
````

### B.4 La variable de flux sur la ligne

On définit la limite de la ligne $|f_\ell| \leq F_\ell$ en bornant la variable de flux entre $-F_\ell$ et $+F_\ell$. Convention : un flux **positif** va du Mozambique vers l'Afrique du Sud.

````python
# TODO : créer la variable de flux f entre -transmission et +transmission.
### DEBUT SOLUTION
#  - m.add_variables(lower=-transmission, upper=transmission, name="flow_MZ_SA") ;
#  - c'est une variable scalaire (une seule ligne), sans coords.
f = m.add_variables(lower=-transmission, upper=transmission, name="flow_MZ_SA")
f
### FIN SOLUTION
````

### B.5 Les contraintes d'équilibre nodal (KCL)

La contrainte d'équilibre unique est remplacée par la loi des nœuds (KCL) : il faut tenir compte de la production locale **et** des flux entrants ou sortants :
$$\sum_s g_{i,s} - \sum_\ell K_{i\ell} f_\ell = d_i$$


````python
# TODO : ajouter une contrainte KCL par pays.
#  - name=f"{country}_KCL".
### DEBUT SOLUTION
#  - boucle for country in countries ; sign = -1 pour "Mozambique", +1 sinon (matrice d'incidence minimale) ;
#  - g.loc[country].sum() + sign * f == load[country] ;
for country in countries:
    sign = -1 if country == "Mozambique" else 1  # minimal incidence matrix
    m.add_constraints(
        g.loc[country].sum() + sign * f == load[country],
        name=f"{country}_KCL",
    )
### FIN SOLUTION
````

````python
# TODO : afficher la contrainte du Mozambique (m.constraints["Mozambique_KCL"]) pour vérifier son écriture.
### DEBUT SOLUTION
m.constraints["Mozambique_KCL"]
### FIN SOLUTION
````

### B.6 L'objectif

$$\min_{g_{i,s}, f_\ell} \sum_{i,s} o_{i,s} g_{i,s}$$
Le produit `g * marginal_costs` s'étend automatiquement sur la dimension `generators` (aligné par les étiquettes) ; `.sum()` additionne sur tous les pays et toutes les technologies.

````python
# TODO : définir l'expression du coût total obj = (g * marginal_costs).sum() et l'afficher.
### DEBUT SOLUTION
obj = (g * marginal_costs).sum()
obj
### FIN SOLUTION
````

````python
# TODO : ajouter l'objectif au modèle (minimisation).
### DEBUT SOLUTION
m.add_objective(obj, sense="min")
### FIN SOLUTION
````

### B.7 Résoudre et lire les résultats

````python
# TODO : résoudre le modèle (m.solve()).
### DEBUT SOLUTION
m.solve()
### FIN SOLUTION
````

````python
# TODO : afficher le coût total optimal.
### DEBUT SOLUTION
m.objective.value
### FIN SOLUTION
````

````python
# TODO : afficher le dispatch optimal de chaque pays et chaque technologie (g.solution.to_dataframe()).
### DEBUT SOLUTION
g.solution.to_dataframe()
### FIN SOLUTION
````

````python
# TODO : afficher la valeur duale de la contrainte d'équilibre sud-africaine (le prix en Afrique du Sud).
### DEBUT SOLUTION
m.constraints["South_Africa_KCL"].dual
### FIN SOLUTION
````

````python
# TODO : afficher la valeur duale de la contrainte d'équilibre mozambicaine (le prix au Mozambique).
### DEBUT SOLUTION
m.constraints["Mozambique_KCL"].dual
### FIN SOLUTION
````

3. Comparez le coût total avec celui de la partie A : pourquoi est-il plus faible ?
4. Que se passerait-il pour les prix si la capacité de la ligne passait à 1 000 MW ?

**Lien avec le cours :** (chapitre 5) contrainte de capacité de transport et congestion ; l'écart de prix entre deux zones est lié à la valeur duale des limites de ligne (voir aussi le TD court terme).

**Éléments de réponse** *(cellule de correction)*

1. Le flux vaut 500 MW, égal à la capacité : la ligne est saturée. Il est positif : l'énergie va du Mozambique vers l'Afrique du Sud.
2. Prix Afrique du Sud = 60 EUR/MWh (gaz marginal) ; prix Mozambique = 0 EUR/MWh (hydraulique excédentaire qui ne peut pas être exportée davantage). L'écart de prix traduit la congestion de la ligne.
3. Le coût total vaut 1 260 000 EUR/h contre 1 290 000 : 500 MW d'hydraulique gratuite remplacent 500 MW de gaz (500 x 60 = 30 000 EUR/h économisés ; le gaz passe de 4 000 à 3 500 MW).
4. L'hydraulique du Mozambique ne peut exporter que 550 MW (1 200 - 650 de demande locale) : avec une capacité supérieure à 550 MW, la ligne n'est plus saturée, les deux prix s'égalisent à 60 EUR/MWh et le coût n'évolue plus.

**À retenir.** Avec linopy, vous écrivez directement les variables de décision, les contraintes physiques et la fonction objectif. La formulation est transparente, mais vous gérez vous-même les bilans nodaux, les signes des flux et les limites du réseau. C'est précisément ce que PyPSA automatise dans la partie 2 : on décrit le réseau (bus, générateurs, charges, lignes) et PyPSA construit toutes ces contraintes.

# Partie 2 — PyPSA

PyPSA propose une représentation structurée d’un système énergétique : un réseau (`Network`) rassemble des nœuds (`Bus`), des générateurs (`Generator`), des consommations (`Load`) et des éléments de transport (`Line`). À partir de ces données, `n.optimize()` construit et résout le problème d’optimisation.

Nous reprenons un exemple à deux zones — Afrique du Sud et Mozambique — et nous nous arrêtons avant l’exemple plus réaliste utilisant le réseau SciGRID.

<img src="https://docs.pypsa.org/latest/assets/logo/logo-primary-light.svg#only-light" width="300px" />

PyPSA signifie **Python for Power System Analysis**.

PyPSA est une librairie Python pour simuler et optimiser les systèmes énergétiques modernes. Il gère notamment :

- des centrales conventionnelles avec engagement des unités (montée et descente en puissance, démarrage, arrêt) ;
- une production éolienne et solaire variable dans le temps ;
- du stockage d'énergie avec pertes de rendement, et des apports et déversements pour l'hydroélectricité ;
- le couplage avec d'autres secteurs énergétiques (électricité, transport, chaleur, industrie) ;
- la conversion entre vecteurs énergétiques (par exemple de l'électricité vers l'hydrogène) ;
- des réseaux de transport (courant alternatif, continu, autres combustibles).

PyPSA s'utilise pour de nombreux types de problèmes (modélisation de marchés de l'électricité, planification d'investissement à long terme, planification de l'extension du réseau de transport) et est conçu pour passer à l'échelle sur de grands réseaux et de longues séries temporelles.

Par rapport à la construction « à la main » d'un système électrique avec `linopy`, PyPSA se charge de :

- gérer les données d'entrée dans un format standardisé ;
- construire le problème d'optimisation standardisé ;
- communiquer avec le ou les solveurs ;
- récupérer et traiter les résultats de l'optimisation ;
- gérer les données de sortie dans un format standardisé.

La [documentation](https://docs.pypsa.org/) le décrit ainsi :

> PyPSA est un cadre Python pour optimiser et simuler les systèmes électriques et énergétiques modernes, avec notamment des centrales conventionnelles avec engagement des unités, une production éolienne et solaire variable, de l'hydroélectricité, du stockage inter-temporel, le couplage avec d'autres secteurs énergétiques, des demandes élastiques et un calcul de répartition des flux linéarisé avec approximation des pertes en réseaux continus et alternatifs. PyPSA est conçu pour passer à l'échelle sur de grands réseaux et de longues séries temporelles. Il s'adresse aux chercheurs, planificateurs et exploitants ayant des bases en programmation qui ont besoin d'un outil rapide, facile à utiliser et transparent pour l'analyse des systèmes électriques et énergétiques.

## De données structurées à l'optimisation

**Le principe de conception de PyPSA est que chaque composant est associé à un ensemble de variables et de contraintes ajoutées au modèle d'optimisation à partir des données saisies pour ce composant.**

Pour une simulation de marché de l'électricité *horaire*, PyPSA résout un problème d'optimisation de la forme suivante :

\begin{equation}
\min_{g_{i,s,t}; f_{\ell,t}; g_{i,r,t,\text{charge}}; g_{i,r,t,\text{discharge}}; e_{i,r,t}} \sum_s o_{s} g_{i,s,t}
\end{equation}
sous les contraintes :
\begin{align}
0 & \leq g_{i,s,t} \leq \hat{g}_{i,s,t} G_{i,s}  & \text{limites de production : générateur} \\
-F_\ell &\leq f_{\ell,t} \leq F_\ell & \text{limites de transmission : ligne}  \\
d_{i,t} &= \sum_s g_{i,s,t} + \sum_r g_{i,r,t,\text{discharge}} - \sum_r g_{i,r,t,\text{charge}} - \sum_\ell K_{i\ell} f_{\ell,t} & \text{KCL : bus} \\
 0 &=\sum_\ell C_{\ell c} x_\ell f_{\ell,t} & \text{KVL : cycles} \\
0 & \leq g_{i,r,t,\text{discharge}} \leq G_{i,r,\text{discharge}}& \text{limites de décharge : stockage} \\
    0 & \leq g_{i,r,t,\text{charge}} \leq G_{i,r,\text{charge}} & \text{limites de charge : stockage} \\
    0 & \leq e_{i,r,t} \leq E_{i,r} & \text{limites d'énergie : stockage} \\
    e_{i,r,t} &= \eta^0_{i,r,t} e_{i,r,t-1} + \eta^1_{i,r,t}g_{i,r,t,\text{charge}} -  \frac{1}{\eta^2_{i,r,t}} g_{i,r,t,\text{discharge}} & \text{consistance : stockage} \\
    e_{i,r,0} & = e_{i,r,|T|-1}  & \text{cyclicité : stockage}
\end{align}

**Variables de décision :**

- $g_{i,s,t}$ : production du générateur au nœud $i$, de technologie $s$, au pas de temps $t$ ;
- $f_{\ell,t}$ : flux de puissance sur la ligne $\ell$ ;
- $g_{i,r,t,\text{dis-/charge}}$ : charge et décharge du stockage $r$ au nœud $i$ et au pas de temps $t$ ;
- $e_{i,r,t}$ : état de charge du stockage $r$ au nœud $i$ et au pas de temps $t$.

**Paramètres :**

- $o_{i,s}$ : coût marginal de production de la technologie $s$ au nœud $i$ ;
- $x_\ell$ : réactance de la ligne $\ell$ ;
- $K_{i\ell}$ : matrice d'incidence ;
- $C_{\ell c}$ : matrice des cycles ;
- $G_{i,s}$ : capacité nominale du générateur de technologie $s$ au nœud $i$ ;
- $F_{\ell}$ : capacité de la ligne $\ell$ ;
- $E_{i,r}$ : capacité énergétique du stockage $r$ au nœud $i$ ;
- $\eta^{0/1/2}_{i,r,t}$ : rendements de maintien (0), de charge (1) et de décharge (2).

## Contraintes globales pour limiter les émissions

Dans l'exemple précédent, il se trouve qu'on dispose d'une capacité de gaz inutilisée, moins carbonée que les centrales à charbon et au pétrole. On pourrait s'en servir pour réduire les émissions du système, mais ce serait plus coûteux. On peut mettre en œuvre une limite d'émissions de dioxyde de carbone sous forme de contrainte.

Dans PyPSA, on utilise pour cela des **contraintes globales (Global Constraints)**, qui ajoutent des contraintes portant sur de nombreux composants à la fois.

Mais il faut d'abord calculer le niveau actuel d'émissions pour fixer une limite raisonnable.

On calcule les émissions de chaque générateur (en tonnes de CO$_2$) ainsi :

$$\frac{g_{i,s,t} \cdot \rho_{i,s}}{\eta_{i,s}}$$

où $\rho$ désigne les émissions spécifiques (tonnes/MWh thermique) et $\eta$ le rendement de conversion (MWh électrique / MWh thermique) du générateur de production $g$ (MWh électrique) :

````python
# Émissions de chaque générateur (t de CO2) = production / rendement x émissions spécifiques du vecteur
e = (
    n.generators_t.p
    / n.generators.efficiency
    * n.generators.carrier.map(n.carriers.co2_emissions)
)
e
````

En sommant, on obtient les émissions totales en tonnes :

````python
# Émissions totales (somme sur les générateurs et les pas de temps)
e.sum().sum()
````

Supposons maintenant qu'on veuille réduire les émissions de 10 % :

````python
# Une contrainte globale : somme des émissions de tous les composants <= constante
# carrier_attribute="co2_emissions" : PyPSA utilise l'attribut co2_emissions des vecteurs
n.add(
    "GlobalConstraint",
    "emission_limit",
    carrier_attribute="co2_emissions",
    sense="<=",
    constant=e.sum().sum() * 0.9,  # 90 % des émissions initiales
)
````

````python
n.optimize(log_to_console=False)
````

````python
# Nouveau dispatch : moins de charbon et de pétrole, plus de gaz
n.generators_t.p
````

````python
n.generators_t.p / n.generators.p_nom
````

````python
# mu : valeur duale de la contrainte d'émissions = prix implicite du CO2 (négatif pour une contrainte <=)
n.global_constraints.mu
````

Peut-on réduire encore les émissions, par exemple de 5 points de plus ?

````python
# On vise 85 % des émissions initiales (et non 0,85 tonne !)
n.global_constraints.loc["emission_limit", "constant"] = e.sum().sum() * 0.85
````

````python
n.optimize(log_to_console=False)
````

**Non !** Sans capacité supplémentaire, on a épuisé les options de réduction des émissions pour cette heure. Le solveur indique que le problème est *infaisable* : il n'existe aucune solution qui satisfasse toutes les contraintes. Il vaut mieux annuler cette modification :

````python
# On revient à la contrainte à 90 % des émissions initiales
n.global_constraints.loc["emission_limit", "constant"] = e.sum().sum() * 0.9
````

> - Modifier une valeur = écrire dans le tableau avec `.loc[nom, attribut]`.

## Partie A — Construire et explorer un réseau pas à pas

### A.1 Construire un réseau à un seul bus (l'Afrique du Sud)

On commence par un réseau minimal : un bus, les quatre centrales sud-africaines, une charge. Le coût marginal d'une centrale thermique se calcule avec le coût du combustible et le rendement : $\text{coût marginal} = \text{coût combustible} / \text{rendement}$.

````python
# TODO : construire le réseau n_sa et l'optimiser.
#  - créer un réseau vide ;
#  - ajouter un bus "SA" (South Africa) ;
#  - ajouter les générateurs dans une boucle
#  - ajouter la demande
#  - optimiser avec n_sa.optimize(solver_name="highs", log_to_console=False).
### DEBUT SOLUTION
# Étape 1 : le réseau, conteneur vide de tous les composants
n_sa = pypsa.Network()

# Étape 2 : un bus, point de raccordement unique des centrales et de la charge
n_sa.add("Bus", "SA")

# Étape 3 : un générateur par technologie. n.add(type, nom, attributs...) :
#  - bus : bus de raccordement ; carrier : technologie ; p_nom : puissance installée (MW) ;
#  - marginal_cost : coût variable en EUR/MWh électrique = coût du combustible / rendement
for tech, p_nom in power_plants["SA"].items():
    n_sa.add(
        "Generator",
        f"SA {tech}",
        bus="SA",
        carrier=tech,
        p_nom=p_nom,
        efficiency=efficiency.get(tech, 1),                                    # 1 pour le vent
        marginal_cost=fuel_cost.get(tech, 0) / efficiency.get(tech, 1),       # 0 pour le vent
    )

# Étape 4 : la charge (puissance consommée, en MW)
n_sa.add("Load", "SA demand", bus="SA", p_set=loads["SA"])

# Étape 5 : construction du problème, résolution, et rapatriement des résultats dans n_sa
statut, condition = n_sa.optimize(solver_name="highs", log_to_console=False)
print("Statut :", statut, "| condition :", condition)
### FIN SOLUTION
````

### A.2 Lire les résultats

Les données statiques (paramètres) sont dans `n_sa.generators`, les résultats qui varient dans le temps dans `n_sa.generators_t.p` (production) et `n_sa.buses_t.marginal_price` (prix de chaque bus). Le seul pas de temps s'appelle `now`.

````python
# TODO : afficher (1) le tableau des générateurs (colonnes bus, carrier, p_nom, marginal_cost),
# (2) la production de chaque générateur, (3) le prix marginal du bus et (4) le coût total de l'objectif.
### DEBUT SOLUTION
# (1) Paramètres : double crochet = liste de colonnes à garder
display(n_sa.generators[["bus", "carrier", "p_nom", "marginal_cost"]])

# (2) Production optimale en MW : lignes = pas de temps, colonnes = générateurs
display(n_sa.generators_t.p)

# (3) Prix marginal du bus : valeur duale de l'équilibre offre-demande, en EUR/MWh
display(n_sa.buses_t.marginal_price)

# (4) Coût total d'exploitation (valeur de la fonction objectif)
print("Objectif :", round(n_sa.objective), "EUR/h")
### FIN SOLUTION
````

## Bilan du TD

- **Linopy** : le modèle d'optimisation est écrit explicitement ; les variables, les contraintes, l'objectif et les valeurs duales sont directement accessibles.
- **PyPSA** : le système est décrit à l'aide de composants physiques (`Bus`, `Generator`, `Load`, `Line`...) ; PyPSA construit le modèle d'optimisation sous-jacent et restitue les résultats dans des tableaux dédiés (`n.generators_t.p`, `n.buses_t.marginal_price`).
- Dans les deux approches, les capacités de production, les contraintes de transport et les bilans énergétiques déterminent le dispatch et les prix marginaux : le prix marginal est la valeur duale de l'équilibre offre-demande.
- Côté syntaxe PyPSA : `n.add()` pour créer, `n.<composants>` pour lire les paramètres, `n.<composants>_t` pour lire les résultats, `.loc[nom, attribut]` pour modifier, `n.copy()` pour comparer des variantes, `n.model` pour examiner le modèle `linopy`.
